applyInPandas() is designed to apply custom Pandas logic to grouped data. It is generally faster than a normal Python UDF because it processes groups as Pandas DataFrames, but it is usually slower than Spark's built-in functions due to the overhead of converting data between Spark and Pandas.

```python
import pandas as pd

data = [
    (1, "IT", 50000),
    (2, "IT", 60000),
    (3, "HR", 40000),
    (4, "HR", 45000),
    (5, "Sales", 70000)
]

df = spark.createDataFrame(
    data,
    ["EmpID", "Department", "Salary"]
)
```

```python
# Function
def add_bonus(pdf):
    pdf["Bonus"] = pdf["Salary"] * 0.10
    return pdf
```

```python
# Output Schema
schema = """
EmpID long,
Department string,
Salary long,
Bonus double
"""
```

```python
result = df.groupBy("Department").applyInPandas(
    add_bonus,
    schema
)
```
