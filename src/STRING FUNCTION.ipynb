```python
from pyspark.sql.functions import *

data = [
    (101, "sona shek", "it", "  Bangalore  ", "sona@gmail.com"),
    (102, "RAHUL", "HR", "Chennai", "rahul@gmail.com"),
    (103, "Priya", "finance", "Mumbai", "priya@gmail.com"),
    (104, "arjun kumar", "IT", " Hyderabad ", "arjun@gmail.com")
]

columns = ["Emp_ID","Name","Department","City","Email"]

df = spark.createDataFrame(data, columns)

df.display()
```

1. upper()

Converts all characters to uppercase. same syntax for lower


```python
df.withColumn("Upper_Name", upper("Name")).display()
```

initcap()

Capitalizes the first letter of every word.

```python
df.withColumn("Proper_Name", initcap("Name")).display()
```

length()

Returns the number of characters in a string.

```python
df.withColumn("Length", length("Name")).display()
```

trim()

Removes spaces from both sides.

```python
df.withColumn("Trim_City", trim("City")).display()
```

ltrim()

Removes spaces only from the left.

rtrim()

Removes spaces only from the right.

```python
df.withColumn("Left_Trim", ltrim("City")).display()
```

concat()

Joins two or more columns without a separator.

```python
df.withColumn( "Details",concat(col("Name"), col("Department"))).display()
```

concat_ws()

Joins columns using a separator.

```python
df.withColumn("Employee",concat_ws(" - ", col("Name"), col("Department"))).display()
```

```python
df.withColumn("Short_Name",substring("Name",1,4)).display()
```

startswith()

Checks whether a string starts with the given prefix.

```python
df.filter(col("Email").startswith("sona")).display()
```

endswith()

Checks whether a string ends with the given suffix.

```python
df.filter(col("Email").endswith(".com")).display()
```

like()

Filters data using SQL wildcard patterns.

% → any number of characters
_ → exactly one character

```python
df.filter(col("Name").like("s%")).display()
```

rlike()

Filters strings using regular expressions.

```python
df.filter(col("Email").rlike("gmail")).display()
```

reverse()

Reverses the characters in a string.

```python
df.withColumn("Reverse",reverse("Name")).display()
```

split()

Splits a string into an array.

```python
df.withColumn( "Words",split("Name"," ")).display()
```

explode()

Converts each array element into a separate row.

```python
df.withColumn("Words",split("Name"," ")).select("Emp_ID",explode("Words").alias("Word")).display()
```

replaces

 an entire value in one or more columns with another value.

Parameters
to_replace → The existing value to search for.
value → The new value that will replace it.
subset (optional) → The column(s) where Spark should perform the replacement.

```python
df.replace("sona shek","anos",subset=["Name"]).display()
```

regexp_replace()

regexp_replace() replaces a part of a string (or text matching a regular expression) with another string.

Unlike replace(), it does not require the entire cell value to match.

regexp_replace(column, pattern, replacement)

```python
from pyspark.sql.functions import regexp_replace

df.withColumn( "New_Name",regexp_replace("Name", "anos", "sona")).display()
```
