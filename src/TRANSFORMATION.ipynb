JOIN

```python
emp_data = [
    (101, "John", 1),
    (102, "Alice", 2),
    (103, "Bob", 3),
    (104, "David", 4)
]

emp_columns = ["emp_id", "emp_name", "dept_id"]

emp_df = spark.createDataFrame(emp_data, emp_columns)

emp_df.display()
```

```python
dept_data = [
    (1, "IT"),
    (2, "HR"),
    (3, "Finance"),
    (5, "Marketing")
]

dept_columns = ["dept_id", "department"]

dept_df = spark.createDataFrame(dept_data, dept_columns)

dept_df.display()
```

```python
emp_df.join(  dept_df,  emp_df.dept_id == dept_df.dept_id,  "inner" ).display()
```

```python
emp_df.join(
    dept_df,
    emp_df.dept_id == dept_df.dept_id,
    "left"
).display()
```

```python
emp_df.join(dept_df,emp_df.dept_id == dept_df.dept_id,"right").display()
```

```python
emp_df.join( dept_df,'dept_id',"right").display()
```

```python
emp_df.join(dept_df,'dept_id',"full").display()
```

A Left Semi Join returns only the rows from the left DataFrame that have a matching row in the right DataFrame.

```python
emp_df.join(dept_df,'dept_id',"left_semi").display()
```

A Left Anti Join returns only the rows from the left DataFrame that do NOT have a match in the right DataFrame.

```python
emp_df.join(dept_df,'dept_id',"left_anti").display()
```

A Self Join is when a DataFrame is joined with itself.

```python
data = [
    (1, "John", None),
    (2, "Alice", 1),
    (3, "Bob", 1),
    (4, "David", 2),
    (5, "Eva", 2),
    (6, "Sam", 3)
]

columns = ["emp_id", "emp_name", "manager_id"]

emp_df = spark.createDataFrame(data, columns)

emp_df.display()
```

```python
emp = emp_df.alias("emp")
mgr = emp_df.alias("mgr")
```

```python
result = emp.join(mgr, col("emp.manager_id") == col("mgr.emp_id"),"left").select('emp_name','manager_id)')

result.display()
```
