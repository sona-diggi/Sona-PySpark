A Transformations notebook should contain only reusable DataFrame transformation functions.

It should not:

❌ Read CSV/Parquet/Delta files
❌ Create Spark sessions
❌ Display data (show())
❌ Write data to tables
❌ Execute ETL logic

Its only job is:

Take a Spark DataFrame → Transform it → Return a Spark DataFrame

```python
from pyspark.sql.functions import *
```

```python
def clean_employee(df):

    df = df.na.drop()

    df = df.dropDuplicates()

    df = df.withColumn(
        "Name",
        upper(col("Name"))
    )

    return df
```
