```python
data = [
    (101, "Sona", "IT", 55000),
    (102, "Rahul", "HR", 48000),
    (103, "Priya", "Finance", 62000)
]

columns = ["Emp_ID", "Name", "Department", "Salary"]

df = spark.createDataFrame(data, columns)

df.display()
```

Mode 1: overwrite--->If the destination already exists, Spark deletes the old data and writes the new data.

```python
df.write.mode("overwrite").csv(
    "/Volumes/workspace/sona/s1/employee_csv"
)
```

Mode 2 : append -->If the destination already exists, don't delete the existing data. Instead, add the new data to it.

```python
new_data = [
    (103, "Priya", "Finance", 62000),
    (104, "Arjun", "IT", 70000)
]

new_df = spark.createDataFrame(new_data, columns)

new_df.display()
```

```python
new_df.write.mode("append").csv("/Volumes/workspace/sona/s1/employee_csv/")
```

```python
new_df.display()
```

```python
df = spark.read.csv("/Volumes/workspace/sona/s1/employee_csv")

df.display()
```

Mode 3: ignore-->If the destination already exists, Spark does nothing.It:

❌ Doesn't overwrite the existing data.
❌ Doesn't append new data.
✅ Simply ignores the write request.

```python
new_df.write.mode("ignore").csv(
    "/Volumes/workspace/sona/s1/employee_csv"
)
```

Mode 4: error / errorifexists

Note: error and errorifexists are the same. errorifexists is just the full name.

If the destination already exists, Spark throws an exception and stops the job.

It:

❌ Doesn't overwrite.
❌ Doesn't append.
❌ Doesn't ignore.
✅ Throws an error.

```python
new_df.write.mode("error").csv(
    "/Volumes/workspace/sona/s1/employee_csv"
)
```
