```python
from pyspark.sql import SparkSession

data = [
    (1, "IT", 50000),
    (2, "IT", 60000),
    (3, "IT", 70000),
    (4, "IT", 80000),
    (5, "IT", 90000),
    (6, "HR", 40000),
    (7, "Sales", 45000)
]

columns = ["EmpID", "Department", "Salary"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
from pyspark.sql.functions import rand, floor, concat, lit, col
```

```python
df1 = df.withColumn(
    "Salt",
    floor(rand() * 3)
)

df1.show()
```

```python
df2 = df1.withColumn(
    "Salted_Department",
    concat(col("Department"), lit("_"), col("Salt"))
)

df2.show(truncate=False)
```

```python
partial = df2.groupBy("Salted_Department").sum("Salary")

partial.show()
```

```python
from pyspark.sql.functions import split
```

```python
partial1 = partial.withColumn(
    "Department",
    split(col("Salted_Department"), "_")[0]
)

partial1.show()
```

```python
final = partial1.groupBy("Department").sum("sum(Salary)")

final.show()
```
