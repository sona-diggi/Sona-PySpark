```python
data = [
    (101, "Sona", "IT", 55000),
    (102, "Rahul", "IT", 65000),
    (103, "Priya", "IT", 60000),
    (104, "Arjun", "HR", 45000),
    (105, "Meena", "HR", 50000),
    (106, "Kiran", "Finance", 70000),
    (107, "Divya", "Finance", 68000),
    (108, "Anil", "Finance", 70000)
]

columns = ["Emp_ID", "Name", "Department", "Salary"]

df = spark.createDataFrame(data, columns)

df.display()
```

```python
from pyspark.sql.window import Window 
from pyspark.sql.functions import *
```

```python
df.withColumn("row_number", row_number().over(Window.partitionBy("Department").orderBy(desc("Salary")))).display()
```

```python
df.withColumn("Rank",rank().over(Window.partitionBy("Department").orderBy(col("Salary").desc()))).filter(col("Rank")==3).display()
```

```python
df.withColumn("Dense Rank",dense_rank().over(Window.partitionBy("Department").orderBy(col("Salary").desc()))).filter(col("Dense Rank")==3).display()
```

NTILE

```python
df.withColumn("ntile",ntile(3).over(Window.partitionBy("Department").orderBy(col("Salary").desc()))).withColumn("Position",when(col("ntile")==1,"high salary").when(col("ntile")==2,"medium salary").otherwise("low salary")).display()
```

LAG-->"Give me the value from the previous row."

```python
df.withColumn("previous salary",lag("salary",1,1).over(Window.orderBy(col("Salary").desc()))).display()
```

```python
df.withColumn("next salary",lead("salary",1,1).over(Window.orderBy(col("Salary").desc()))).display()
```

lead() returns the value from the next row within the same window.

SUM

running sum

```python
df.withColumn("summ",sum("salary").over(Window.partitionBy("Department").orderBy(col("Salary").desc()).rowsBetween(Window.unboundedPreceding,Window.currentRow))).display()
```
